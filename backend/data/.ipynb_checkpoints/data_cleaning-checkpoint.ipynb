```python
import pandas as pd
import zipfile
```

We can probably remove serving size and tags.

```python
recipes = pd.read_csv("recipes_w_search_terms.csv")
recipes.head()
```

```python
# cleaned = recipes.drop(["id","serving_size","tags"], axis=1, inplace=False)
cleaned = recipes.drop(["serving_size","tags"], axis=1, inplace=False)
cleaned = cleaned[(cleaned['name'].str.count(' ') < 3) & (cleaned['description'].notna())]
cleaned.head()
```

```python
ratings = pd.read_csv('RAW_interactions.csv')
ratings.head()
```

```python
avg_ratings = ratings.groupby('recipe_id')['rating'].mean().reset_index()
avg_ratings = avg_ratings.rename(columns={'rating': 'average_rating'})
avg_ratings = avg_ratings.rename(columns={'recipe_id': 'id'})
avg_ratings.head()
```

```python
combined_data = pd.merge(
    cleaned,
    avg_ratings,
    on='id',
    how='left'
)

combined_data.head()
```

```python
combined_data.to_json('recipes_cleaned.json', orient='records', indent=4)
```

```python
names = pd.read_json("recipes_cleaned.zip")
names = names['name']
names = names.apply(lambda x: x.lower())
print(names.head())
print(len(names))
names.to_csv('recipes_names.csv', index=False)
```

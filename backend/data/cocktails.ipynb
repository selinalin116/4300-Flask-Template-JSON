```python
import requests
import json
```

```python
def fetch_cocktails():
    """
    Fetch cocktails from TheCocktailDB API.
    """
    cocktails = []
    for letter in 'abcdefghijklmnopqrstuvwxyz':
        response = requests.get(f'https://www.thecocktaildb.com/api/json/v1/1/search.php?f={letter}')
        if response.ok and response.json().get('drinks'):
            cocktails.extend(response.json()['drinks'])
    return cocktails
```

```python
cocktails = fetch_cocktails()

# Optional: print how many you got
print(f"Fetched {len(cocktails)} cocktails.")
```

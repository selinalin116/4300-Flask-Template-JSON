```python
import pandas as pd
import json

PAIRING_CSV_PATH = "TasteTrios - Sheet1.csv"
pairing_df = pd.read_csv(PAIRING_CSV_PATH).dropna()

# Clean ingredient strings
pairing_df["Ingredient 1"] = pairing_df["Ingredient 1"].str.lower().str.strip()
pairing_df["Ingredient 2"] = pairing_df["Ingredient 2"].str.lower().str.strip()
pairing_df["Ingredient 3"] = pairing_df["Ingredient 3"].str.lower().str.strip()

# Build dictionary with string keys (sorted trios)
compatible_trios = {
    ", ".join(sorted([row["Ingredient 1"], row["Ingredient 2"], row["Ingredient 3"]])): row["Classification Output"]
    for _, row in pairing_df.iterrows()
}

# Save to JSON
with open("taste_trios.json", "w") as f:
    json.dump(compatible_trios, f, indent=2)

print("✅ Saved as efficient_taste_trios.json")
```
